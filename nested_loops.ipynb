```python
rows = int(input("Enter the number of rows: "))
sign = input("Enter the sign: ")

for x in range(rows):
    for y in range(x + 1):
        print(sign, end="")
    print()
```

```python
rows = int(input("Enter the number of rows: "))
sign = input("Enter the sign: ")

for x in reversed (range(rows)):
    for y in range(x + 1):
        print(sign, end="")
    print()
```
