Validate user input exercise


```python
#1 Username is no more than 12 characters

username = input("Enter your username: ")

limit = (len(username) <= 12) and username.isalpha() and not username.isdigit()

if limit == True:
    print("Username valid")
    
else:
    print("Username invalid")
```

```python
#2 Username must not contain spaces

username2 = input("Enter your username: ")

rule = username2.isalpha()

if rule == True:
    print("No spaces")

else:
    print("remove space")
```

```python
#3 Username must not contain digits

username3 = input("Enter your username: ")

rule = username3.isdigit()

if rule == False:
    print("No digits")

else:
    print("remove digits")
```

```python
username = input("Enter your username: ")

if len(username) > 12:
    print("Your username is too big")

elif not username.find(" ") == -1:
    print("Your username can't contain spaces")

elif username.isdigit() == True:
    print("Your username can't contain digits")

else:
    print("Username valid")
```
