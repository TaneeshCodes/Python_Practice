Dictionary: A collection of {key:value} pairs. (ordered and changeable) No Duplicates

```python
capitals = {"USA": "Washington D.C",
            "India": "New Delhi",
            "China": "Beijing",
            "Russia": "Moscow"}

print(capitals.get("USA"))

capitals.update({"Germany": "Berlin"})

print(capitals)

if capitals.get("Germany"):
    print("Capital Exists")

else:
    print("Capital Doesn't Exist")
```

```python
capitals = {"USA": "Washington D.C",
            "India": "New Delhi",
            "China": "Beijing",
            "Russia": "Moscow"}

capitals.pop("China") # Removes china
capitals.popitem() # Removes latest country i.e Russia in this case
# capitals.clear() # Clears everything

print(capitals)
```

```python
capitals = {"USA": "Washington D.C",
            "India": "New Delhi",
            "China": "Beijing",
            "Russia": "Moscow"}

keys = capitals.keys()
for key in capitals.keys():
    print(key)
```

```python
capitals = {"USA": "Washington D.C",
            "India": "New Delhi",
            "China": "Beijing",
            "Russia": "Moscow"}

values = capitals.values()
for value in capitals.values():
    print(value)
```

```python
capitals = {"USA": "Washington D.C",
            "India": "New Delhi",
            "China": "Beijing",
            "Russia": "Moscow"}

for key, value in capitals.items():
    print(f"{key}: {value}")
```
