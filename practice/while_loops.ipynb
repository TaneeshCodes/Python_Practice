While Loop = Execute some code WHILE some condition remains true

```python
name = input("Enter your name: ")

while name == "":
    print("You did not enter the name")
    name = input("Enter your name: ")
print(f"Hello {name}")
```

```python
age = int(input("Enter your age: "))

while age == "":
    print("Please enter your age")
    age = int(input("Enter your age: "))
print(f"Your age is {age}")
```

```python
age = int(input("Enter your age: "))

while age < 0:
    print("Age can't be negative")
    age = int(input("Enter your age: "))
print(f"Your age is {age}")
```

2 conditions 

```python
age = int(input("Enter your age: "))

while age < 0 or age > 100:

    if age < 0:
        print("Age can't be in negative")

    elif age > 100:
        print("Age can't be greater than 100")

    age = int(input("Enter your age: "))

print(f"You are {age} years old")
```

```python
food = input("Enter a food you like (q to quit): ")

while not food == "q":
    print(f"You like {food}")

    food = input("Enter another food you like (q to quit): ")

print("Bye")
```

```python
num = int(input("Enter the number between 1 to 10"))

while num > 10 or num < 1:

    if num > 10:
        print("Number can't be greater than 10")

    elif num < 1:
        print("Number can't be lesser than 1")

    num = int(input("Enter the number between 1 to 10"))

print(f"Your number is {num}")
```
