Indexing: Accessing elements of a sequence using [] (indexing operators) [start: end: step]

```python
credit_number = "123-456-789"
print(credit_number[0]) #prints index 0 i.e 1
print(credit_number[-1]) #prints last number i.e 9
print(credit_number[-3]) # print third last number i.e 7
print(credit_number[:5]) # prints from index 0 to 5 i.e 123-5
print(credit_number[5:]) #prints from index 5 onwards i.e 56-789
print(credit_number[0 : 5]) # prints from index 0 to 5 i.e 123-4
print(credit_number[5 : 7]) # prints from index 5 to 7 i.e 56
print(credit_number[::2]) #prints numbers in multiples of 2 i.e 134679
print(credit_number[::-1]) #reverse the string
```
