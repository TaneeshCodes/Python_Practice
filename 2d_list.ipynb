2D List are just lists made up of multiple lists

```python
fruits = ["apple", "orange", "lichi"]
vegetable = ["celery", "carrots", "potatoes"]
meats = ["chicken", "fish", "turkey"]

groceries = [fruits, vegetable, meats]
print(groceries)
```

```python
fruits = ["apple", "orange", "lichi"]
vegetable = ["celery", "carrots", "potatoes"]
meats = ["chicken", "fish", "turkey"]

groceries = [fruits, vegetable, meats]
print(groceries[1][2])
```

```python
groceries = [["apple", "orange", "lichi"],
             ["celery", "carrots", "potatoes"],
             ["chicken", "fish", "turkey"]]

for collection in groceries:
    for food in collection:
        print(food, end=" ")
    print()
```

```python
groceries = [["apple", "orange", "lichi"],
             ["celery", "carrots", "potatoes"],
             ["chicken", "fish", "turkey"]]


for food in groceries:
    print(food)
```

```python
num_pad = ((1, 2, 3),
           (4, 5, 6),
           (7, 8, 9),
           ("*", 0, "#"))

for nums in num_pad:
    for number in nums:
        print(number, end=" ")
    print()
```
