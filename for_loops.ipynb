For Loops: Execute a block of code a fixed number of times. You can iterate over a range, string, sequence, etc.

```python
for x in range(1, 11): # Last number i.e 11 in this case is excluded
    print(x)
```

```python
for y in reversed(range(1, 11)):
    print(y)

print("Happy New Year")
```

```python
for x in range(1, 21):
    if x == 13:
        continue    #continue: skips that iteration
    else:
        print(x)
```

```python
for x in range(1, 21):
    if x == 13:
        break    #break: only iterates until the break condition and the loop finishes once condition is reached or satisfied
    else:
        print(x)
```
