```python
import time

user = int(input("For how long u want to set the timer: "))
time.sleep(user)

print(f"we have passed {user} seconds")
```

```python
import time

user = int(input("For how long u want to set the timer: "))
for x in range(0, user):
    print(x)
    time.sleep(1)

print(f"we have passed {user} seconds")
```

```python
import time

user = int(input("For how long u want to set the timer: "))
for x in reversed (range(0, user)):
    print(x)
    time.sleep(1)

print(f"we have passed {user} seconds")
```

Another method to count backwards

```python
import time

user = int(input("For how long u want to set the timer: "))
for x in range(user, 0, -1):
    print(x)
    time.sleep(1)

print(f"we have passed {user} seconds")
```

```python
import time

user = int(input("For how long u want to set the timer: "))

for x in range(user, 0, -1):
    seconds = x % 60
    minutes = int(x / 60) % 60
    hours = int(x / 3600)
    print(f"{hours:02}:{minutes:02}:{seconds:02}")
    time.sleep(1)

print(f"we have passed {user} seconds")
```
