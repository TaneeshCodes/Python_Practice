Collection: Single "Variable" used to store multiple values

List = [] ordered and changeable. Duplicates OK


Set = {} unordered and immutable, but Add/Remove OK. NO duplicates

Tuple = () ordered and unchangeable. Duplicates OK. Faster

```python
fruits = ["Apple", "Orange", "Lichi", "Mango"]
print(fruits[1])
```

```python
fruits = ["Apple", "Orange", "Lichi", "Mango"]
print(fruits[:2])
```

```python
fruits = ["Apple", "Orange", "Lichi", "Mango"]
print(fruits[::-1])
```

```python
fruits = ["Apple", "Orange", "Lichi", "Mango"]

for fruit in fruits:
    print(fruit)
```

```python
fruits = ["Apple", "Orange", "Lichi", "Mango"]

print(len(fruits))
print("Apple" in fruits)

fruits[1] = "Pineapple"  #Reassigned Value

for fruit in fruits:
    print(fruit)
```

```python
fruits = ["Apple", "Orange", "Lichi", "Mango"]
fruits.append("Apple")

for fruit in fruits:
    print(fruit)
```

```python
fruits = ["Apple", "Orange", "Lichi", "Mango"]
fruits.remove("Apple")

for fruit in fruits:
    print(fruit)
```

```python
fruits = ["Apple", "Orange", "Lichi", "Mango"]
fruits.insert(0, "pineapple")

for fruit in fruits:
    print(fruit)
```

```python
fruits = ["Apple", "Orange", "Lichi", "Mango"]
fruits.sort()

for fruit in fruits:
    print(fruit)
```

```python
fruits = ["Apple", "Orange", "Lichi", "Mango"]
fruits.reverse()

for fruit in fruits:
    print(fruit)
```

```python
fruits = ["Apple", "Orange", "Lichi", "Mango"]
fruits.clear()

for fruit in fruits:
    print(fruit)
    
```

```python
fruits = ["Apple", "Orange", "Lichi", "Mango"]
print(fruits.index("Apple"))
```
