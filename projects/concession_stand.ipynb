```python
menu = {"pizza": 3.00,
        "icecream": 1.99,
        "fries": 2.69,
        "burger": 3.50,
        "popcorn": 6.79}

cart = []
total = 0

print("---------- MENU ----------")
for key, value in menu.items():
    print(f"{key:10}: ${value:.2f}")

while True:
    food = input("Select an item (q to quit): ").lower( )
    if food == "q":
        break
    elif menu.get(food) is not None:
        cart.append(food)

print("------ YOUR ORDER --------")

for food in cart:
    total += menu.get(food)
    print(food, end=" ")

print()
print(f"Your total is ${total:.2f}")
```

```python
theatre = {"Spider-Man BND": 10.99,
          "Odyssey": 12.50,
          "Justice League": 5.99,
          "Batman": 20.99,
          "Avengers": 15.60}

movie = []
total = 0

print("----- TICKETS -----")
for key, values in theatre.items():
    print(f"{key:10}: ${values:.2f}")

while True:
    ticket = input("Select your movie (q to quit): ")
    if ticket == "q":
        break
    elif theatre.get(ticket) is not None:
        movie.append(ticket)

print("------ YOUR ORDER --------")

for ticket in movie:
    total += theatre.get(ticket)
    print(ticket, end=" ")

print()
print(f"Your total is ${total}")



```
