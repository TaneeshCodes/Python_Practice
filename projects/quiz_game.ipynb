```python
print("Welcome to KBC")
print()
questions = [
    {
        "question" : "What is the capital of India?",
        "options" : ["A. Mumbai", "B. Delhi", "C. Chennai", "D. Kolkata"],
        "answer" : "B",
        "money" : 1000
    },
    {
        "question": "Which planet is known as the Red Planet?",
        "options": [
            "A. Earth",
            "B. Venus",
            "C. Mars",
            "D. Jupiter"
        ],
        "answer": "C",
        "money": 5000
    },

    {
        "question": "Who developed Python?",
        "options": [
            "A. James Gosling",
            "B. Dennis Ritchie",
            "C. Guido van Rossum",
            "D. Bjarne Stroustrup"
        ],
        "answer": "C",
        "money": 10000
    }
]

for question in questions:
    print(question["question"])
    for option in question["options"]:
        print(option)

    user_ip = input("Enter your answer: ").upper()

    print()
    if user_ip == question["answer"]:
        print(f"Correct, you have won {question['money']}")
    else:
        print("Wrong")
        break

   
```
